```python
import pandas as pd
import numpy as np
```

```python
#1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Add a new column that is the product of the first two columns.

# Create a DataFrame with 3 columns and 5 rows
df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)),
                  columns=['A', 'B', 'C'])

# Add a new column as the product of first two columns
df['Product'] = df['A'] * df['B']

# Display the DataFrame
print(df)
```

```python
#2. Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. Compute the row-wise and column-wise sum.

df = pd.DataFrame(np.random.randint(1, 100, size=(4, 3)),
                  columns=['A', 'B', 'C'])

# Compute row-wise sum
row_sums = df.sum(axis=1)

# Compute column-wise sum
col_sums = df.sum(axis=0)

print("DataFrame:")
print(df)
print("\nRow-wise sums:")
print(row_sums)
print("\nColumn-wise sums:")
print(col_sums)
```

```python
#3. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)),columns=['A', 'B', 'C'])

# Introduce some NaN values
df.loc[0, 'A'] = np.nan
df.loc[2, 'B'] = np.nan
df.loc[4, 'C'] = np.nan


# Fill NaN values with the mean of the respective columns
df.fillna(df.mean(), inplace=True)

print(df)
```

```python
#4. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.

df = pd.DataFrame(np.random.randint(1, 100, size=(6, 4)), columns=['A', 'B', 'C', 'D'])

# Introduce some NaN values
df.loc[0, 'A'] = np.nan
df.loc[2, 'B'] = np.nan
df.loc[4, 'C'] = np.nan

# Drop rows with any NaN values
df.dropna(inplace=True)

print(df)
```

5. Data Aggregation 

```python
#Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(1, 100, size=10)
})

# Group by 'Category' and compute sum and mean of 'Value'
grouped = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print("DataFrame:")
print(df)   

# Display grouped result
print("\nGrouped Data:")
print(grouped)
```

```python
#6. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

df = pd.DataFrame({
    'Product': np.random.choice(['P1', 'P2', 'P3'], size=10),
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Sales': np.random.randint(1, 100, size=10)
})  

# Group by 'Category' and compute total sales
total_sales = df.groupby('Category')['Sales'].sum()

print("DataFrame:")
print(df)
print("\nTotal Sales by Category:")
print(total_sales)
```

```python
#7. Merging DataFrames

#Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Age': [25, 30, 35]
})

merged_df = pd.merge(df1, df2, on='ID')
print("Merged DataFrame:")
print(merged_df)
```

```python
# 8. MultiIndex DataFrame

# Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.# Create a MultiIndex DataFrame

df = pd.DataFrame({'Value': [1, 2, 3, 4, 5, 6]},
    index=pd.MultiIndex.from_tuples([
        ('A', 1), ('A', 2),
        ('B', 1), ('B', 2),
        ('C', 1), ('C', 2)
    ], names=['Category', 'Subcategory']))

# Display DataFrame
print("MultiIndex DataFrame:")
print(df)

# Access category A
print("\nCategory A:")
print(df.loc['A'])

# Access a specific row
print("\nCategory B, Subcategory 2:")
print(df.loc[('B', 2)])

# Slice categories A and B
print("\nSlicing A and B:")
print(df.loc['A':'B'])

# Access values of subcategory 1
print("\nSubcategory 1:")
print(df.xs(1, level='Subcategory'))
```

```python
#9. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

df = pd.DataFrame({'Value': np.random.randint(1, 10, size=6)},
    index=pd.MultiIndex.from_tuples([
        ('A', 1), ('A', 2),
        ('B', 1), ('B', 2),
        ('C', 1), ('C', 2)
    ], names=['Category', 'Subcategory']))

# Display DataFrame
print("MultiIndex DataFrame:")
print(df)

print("\nSum of Values by Category and Subcategory:")

print(df.groupby(level='Category').sum())
print(df.groupby(level='Subcategory').sum())
```

```python
# 10. Applying Functions

# Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.

df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)), columns=['A', 'B', 'C'])
print("Original DataFrame:")
print(df)
df = df.apply(lambda x: x * 2)
print("\nDataFrame after doubling values:")
print(df)
```

```python
# 11. Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.

df = pd.DataFrame(np.random.randint(1, 100, size=(6, 3)), columns=['A', 'B', 'C'])
# Apply a lambda function to create a new column that is the sum of the existing columns
print("Original DataFrame:")
print(df)   
df['Sum'] = df.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)
print("\nDataFrame after adding 'Sum' column:")
print(df)
```
